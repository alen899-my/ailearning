```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
Imagepath="spiderman.jpg"
```

```python
img=cv2.imread(Imagepath)
```

```python
img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

```python
img_gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image (RGB)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
cv2.imshow("spiderman",img_rgb)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
half=cv2.resize(img_rgb,(0,0),fx=0.1,fy=0.1)
bigger=cv2.resize(img_rgb,(1050,1610))
images=[half,bigger]
for i in range(2):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
plt.show()
```

```python
bg_img=cv2.cvtColor(img_rgb,cv2.COLOR_BGR2GRAY)
```

```python
cv2.imshow("black and white",bg_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
bg_img=cv2.rotate(bg_img,cv2.ROTATE_180_CLOCKWISE)
```

```python
cv2.imshow("black and white",bg_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
edge=cv2.Canny(img_rgb,100,200)
```

```python
cv2.imshow('edges',edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
