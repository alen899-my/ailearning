```python
import cv2
import time
```

```python
haar_cascade='cars.xml'
```

```python
car_Cascade=cv2.CascadeClassifier(haar_cascade)
```

```python
if car_Cascade.empty():
    raise Exception("eroor loading haar cascade file")
```

```python
Video='car-video.mp4'
```

```python
cap=cv2.VideoCapture(Video)
```

```python
if not cap.isOpened():
    raise Exception("eroor loading video")
```

```python
car_detected=False
detection_time=None
```

```python
car_detected = False  
detection_time = None  

while cap.isOpened():
    ret, frames = cap.read()
    if not ret:
        break 

    gray = cv2.cvtColor(frames, cv2.COLOR_BGR2GRAY)
    cars = car_Cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=3)

    
    if len(cars) > 0 and not car_detected:
        car_detected = True
        detection_time = time.time()
        print("Car detected!")

    if car_detected and time.time() - detection_time > 5:  
        break

    for (x, y, w, h) in cars:
        cv2.rectangle(frames, (x, y), (x + w, y + h), (0, 0, 255), 2)

    cv2.imshow('video', frames)

    if cv2.waitKey(33) == 27: 
        break
```

```python
cap.release()  
cv2.destroyAllWindows()
```
