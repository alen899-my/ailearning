```python
# ...existing code...
import os
import torch
from matplotlib import pyplot as plt
from PIL import Image

# Load YOLOv5 model (pre-trained on COCO dataset)
model = torch.hub.load('ultralytics/yolov5', 'yolov5s')  # 'yolov5s', 'yolov5m', 'yolov5l', 'yolov5x'

# Load an image (use full path if image is local)
img = 'C:/Users/ADMIN/note/yolo/spiderman.jpg'  # Replace with your image path

# If local image not found, fallback to a public sample URL
if not os.path.exists(img):
    print(f"Local file '{img}' not found — using sample image URL instead.")
    img = 'https://ultralytics.com/images/zidane.jpg'  # fallback URL

# Inference
results = model(img)

# Show image with bounding boxes
results.show()

# Print detected objects
print(results.pandas().xyxy[0])  # Bounding boxes, confidence, class
# ...existing code...
```
