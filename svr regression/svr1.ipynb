```python
X=data[[ 'tip', 'sex', 'smoker', 'day', 'time', 'size']]
```

```python
y=data["total_bill"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
X_train
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
```

```python
X_train["sex"]=le1.fit_transform(X_train['sex'])
X_train["smoker"]=le2.fit_transform(X_train["smoker"])
X_train["time"]=le3.fit_transform(X_train["time"])
```

```python
X_train
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
X_test["sex"]=le1.transform(X_test['sex'])
X_test["smoker"]=le2.transform(X_test["smoker"])
X_test["time"]=le3.transform(X_test["time"])
```

```python
X_test.head()
```

```python
#onehot encoding---column transformer
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ct=ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),[3])],remainder="passthrough")
```

```python
import sys
import numpy
numpy.set_printoptions(threshold=sys.maxsize)
X_train=ct.fit_transform(X_train)
```

```python
X_test=ct.transform(X_test)
```

```python
X_test
```

```python
#support vector regrssion
from sklearn.svm import SVR
svr=SVR()
```

```python
svr.fit(X_train,y_train)
```

```python
y_pred=svr.predict(X_test)
```

```python
from sklearn.metrics import r2_score ,mean_absolute_error
```

```python
r2_score(y_test,y_pred)
```

```python
#tuning
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
```

```python
param_grid={'C': [0.1,.1,10,100,1000],
    'gamma':[1,0.1,0.01,0.001,0.0001],
    'kernel':['rbf']

}
```

```python
grid=GridSearchCV(SVC(),param_grid=param_grid,cv=5,verbose=3,refit=True)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
from sklearn.svm import SVR
from sklearn.model_selection import GridSearchCV

param_grid = {
    'C': [0.1, 1, 10, 100, 1000],
    'gamma': [1, 0.1, 0.01, 0.001, 0.0001],
    'kernel': ['rbf']
}

grid = GridSearchCV(SVR(), param_grid=param_grid, cv=5, verbose=3, refit=True)
grid.fit(X_train, y_train)  # now y_train is continuous, so this works correctly
```

```python
grid.best_params_
```

```python
grid_pred=grid.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("R² score:", r2_score(y_test, grid_pred))
print("Mean Absolute Error:", mean_absolute_error(y_test, grid_pred))
print("Mean Squared Error:", mean_squared_error(y_test, grid_pred))
```
