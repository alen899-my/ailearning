```python
data = """
Dream, dream, dream. Dreams transform into thoughts and thoughts result in action. 
You have to dream before your dreams can come true. 
Excellence happens not by accident. It is a process. 
You must have a goal, a vision, and work hard with dedication and perseverance. 
Never stop fighting until you arrive at your destined place – that is, the unique you. 
Have an aim in life, continuously acquire knowledge, work hard, and have perseverance to realize the great life.
"""
```

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download("stopwords")
```

```python
stopwords.words('english')
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer=PorterStemmer()
```

```python
sentences=nltk.sent_tokenize(data)
```

```python
##apply stop words and filter anad then apply stemming
```

```python
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[stemmer.stem(word) for word in words if word not in set(stopwords.words("english"))]
    sentences[i]=" ".join(words)
```

```python
sentences
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstemmerr=SnowballStemmer("english")

sentencess=nltk.sent_tokenize(data)
```

```python
for i in range(len(sentencess)):
    words=nltk.word_tokenize(sentencess[i])
    words=[snowballstemmerr.stem(word) for word in words if word not in set(stopwords.words("english"))]
    sentencess[i]=" ".join(words)
```

```python
sentences=nltk.sent_tokenize(data)
```

```python
sentences
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lem=WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[lem.lemmatize(word,pos="v") for word in words if word not in set(stopwords.words("english"))]
    sentences[i]=" ".join(words)
```
