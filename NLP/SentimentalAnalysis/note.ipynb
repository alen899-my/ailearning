```python
# %%
# ===============================
# WORD2VEC FEATURE REPRESENTATION
# ===============================

from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import numpy as np
import nltk

nltk.download('punkt')

# Tokenize reviews
X_train_tokens = [word_tokenize(text) for text in X_train]
X_test_tokens = [word_tokenize(text) for text in X_test]

# %%
# Train Word2Vec model
w2v_model = Word2Vec(sentences=X_train_tokens,
                     vector_size=100,  # you can tune this
                     window=5,
                     min_count=2,
                     workers=4)

# %%
# Function to average Word2Vec embeddings for each document
def get_avg_w2v(tokens, model, vector_size):
    vec = np.zeros(vector_size)
    count = 0
    for word in tokens:
        if word in model.wv:
            vec += model.wv[word]
            count += 1
    if count > 0:
        vec /= count
    return vec

# %%
# Convert train/test tokens to averaged Word2Vec vectors
X_train_w2v = np.array([get_avg_w2v(tokens, w2v_model, 100) for tokens in X_train_tokens])
X_test_w2v = np.array([get_avg_w2v(tokens, w2v_model, 100) for tokens in X_test_tokens])

print("Word2Vec feature shape:", X_train_w2v.shape)

# %%
# TRAIN ML ALGORITHMS ON WORD2VEC FEATURES
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

# Example 1: Random Forest
rf_model_w2v = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model_w2v.fit(X_train_w2v, y_train)

# Example 2: Logistic Regression
lr_model_w2v = LogisticRegression(max_iter=500)
lr_model_w2v.fit(X_train_w2v, y_train)

# %%
# EVALUATION
y_pred_rf = rf_model_w2v.predict(X_test_w2v)
y_pred_lr = lr_model_w2v.predict(X_test_w2v)

print("Random Forest (Word2Vec) Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Logistic Regression (Word2Vec) Accuracy:", accuracy_score(y_test, y_pred_lr))

print("\nClassification Report (Logistic Regression):")
print(classification_report(y_test, y_pred_lr))
```

```python
# %%
# ===============================
# TESTING ON NEW DATA
# ===============================

# Example new reviews
new_reviews = [
    "The product quality is amazing and delivery was fast!",
    "Terrible experience, it broke after two days.",
    "It's okay, not great but not too bad either."
    "good",
    ""
]

# %%
# Apply the same preprocessing steps

def preprocess_text(text):
    text = text.lower()
    text = re.sub('[^a-zA-Z0-9]+', ' ', text)
    text = re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '', text)
    text = BeautifulSoup(text, 'html.parser').get_text()
    text = " ".join(text.split())
    text = " ".join([lematizer.lemmatize(word) for word in text.split()])
    return text

new_reviews_clean = [preprocess_text(r) for r in new_reviews]

# %%
# ====== 1. Bag of Words ======
new_bow = bow.transform(new_reviews_clean).toarray()
bow_preds = nb_model_bow.predict(new_bow)

# ====== 2. TF-IDF ======
new_tfidf = tfidf.transform(new_reviews_clean).toarray()
tfidf_preds = nb_model_tf.predict(new_tfidf)

# ====== 3. Word2Vec ======
new_tokens = [word_tokenize(r) for r in new_reviews_clean]
new_w2v = np.array([get_avg_w2v(tokens, w2v_model, 100) for tokens in new_tokens])
w2v_preds = lr_model_w2v.predict(new_w2v)  # or rf_model_w2v

# %%
# Convert numeric predictions to labels
def label_text(x): 
    return "Positive" if x == 1 else "Negative"

results_df = pd.DataFrame({
    "Review": new_reviews,
    "Prediction_BoW": [label_text(p) for p in bow_preds],
    "Prediction_TFIDF": [label_text(p) for p in tfidf_preds],
    "Prediction_Word2Vec": [label_text(p) for p in w2v_preds]
})

results_df
```
