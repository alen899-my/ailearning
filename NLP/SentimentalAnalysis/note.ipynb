```python
data["rating"]=data['rating'].apply(lambda x:0 if x<3 else 1)
```

```python
data["rating"].unique()
```

```python
#1.lower all the cases
data["reviewText"]=data["reviewText"].str.lower()
```

```python
data.head()
```

```python
#removing speacial characters
```

```python
import re
data["reviewText"]=data["reviewText"].apply(lambda x:re.sub('[^a-zA-z0-9]+',' ',x))
```

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')  # download the stopwords dataset
```

```python
from bs4 import BeautifulSoup
!pip install lxml
```

```python
data['reviewText']=data['reviewText'].apply(lambda x: re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '' , str(x)))
## Remove html tags
data['reviewText'] = data['reviewText'].apply(lambda x: BeautifulSoup(x, 'html.parser').get_text())

## Remove any additional spaces
data['reviewText']=data['reviewText'].apply(lambda x: " ".join(x.split()))
```

```python
data.head( )
```

```python
#lematizer
from nltk.stem import WordNetLemmatizer
```

```python
lematizer=WordNetLemmatizer()
```

```python
def lematize_words(text):
    return " ".join([lematizer.lemmatize(word) for word in text.split()])
```

```python
data["reviewText"]=data["reviewText"].apply(lambda x:lematize_words(x))
```
